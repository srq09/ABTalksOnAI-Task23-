```python
# ============================================================
# ABTalks Season-3 | AI Engineering | Task Day-23
# Feature Engineering Basics - TF-IDF Text Classification
# Professional Original Solution
# ============================================================

import re
import string
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# STEP 1: SAMPLE DATASET
# ============================================================
# Binary sentiment classification dataset

data = {
    "text": [
        "I absolutely loved this product, amazing quality and performance!",
        "Worst purchase ever, completely disappointed.",
        "Excellent customer service and fast delivery.",
        "The item broke after two days, very poor quality.",
        "Highly recommended, worth every penny.",
        "Terrible experience, I want my money back.",
        "Fantastic design and easy to use.",
        "Not good, product does not work as expected.",
        "Superb packaging and premium build quality.",
        "Very frustrating and unreliable product."
    ],
    "label": [
        "Positive",
        "Negative",
        "Positive",
        "Negative",
        "Positive",
        "Negative",
        "Positive",
        "Negative",
        "Positive",
        "Negative"
    ]
}

df = pd.DataFrame(data)

# ============================================================
# STEP 2: TEXT PREPROCESSING FUNCTION
# ============================================================

def clean_text(text):
    """
    Clean raw text for NLP processing
    """
    text = text.lower()                           # lowercase
    text = re.sub(r"http\S+", "", text)          # remove URLs
    text = re.sub(r"\d+", "", text)              # remove numbers
    text = text.translate(str.maketrans("", "", string.punctuation))  # remove punctuation
    text = re.sub(r"\s+", " ", text).strip()    # remove extra spaces
    return text

df["clean_text"] = df["text"].apply(clean_text)

# ============================================================
# STEP 3: TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    df["clean_text"],
    df["label"],
    test_size=0.2,
    random_state=42,
    stratify=df["label"]
)

# ============================================================
# STEP 4: BUILD SKLEARN PIPELINE
# TF-IDF + CLASSIFIER
# ============================================================

pipeline = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            stop_words="english",
            max_features=1000,
            ngram_range=(1, 2)
        )
    ),
    (
        "classifier",
        LogisticRegression(random_state=42)
    )
])

# ============================================================
# STEP 5: TRAIN MODEL
# ============================================================

pipeline.fit(X_train, y_train)

# ============================================================
# STEP 6: PREDICTIONS
# ============================================================

y_pred = pipeline.predict(X_test)

# ============================================================
# STEP 7: EVALUATION
# ============================================================

accuracy = accuracy_score(y_test, y_pred)

print("=" * 60)
print("TF-IDF TEXT CLASSIFICATION REPORT")
print("=" * 60)

print(f"\nAccuracy: {accuracy:.2f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# ============================================================
# STEP 8: FEATURE EXTRACTION INSPECTION
# ============================================================

tfidf_vectorizer = pipeline.named_steps["tfidf"]
features = tfidf_vectorizer.get_feature_names_out()

print("\nTop Extracted TF-IDF Features:")
print(features[:20])

# ============================================================
# STEP 9: TEST CUSTOM INPUT
# ============================================================

custom_text = [
    "Amazing product with fantastic quality and quick delivery",
    "Very bad experience, poor build and waste of money"
]

custom_text_clean = [clean_text(text) for text in custom_text]

predictions = pipeline.predict(custom_text_clean)

print("\nCustom Predictions:")
for text, pred in zip(custom_text, predictions):
    print(f"Text: {text}")
    print(f"Prediction: {pred}")
    print("-" * 50)
```
